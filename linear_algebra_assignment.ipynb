### Q2 Part (i) — Shape Change Idea

**No run code.** Guess what happens when you times math box `A` and lists. Tell **how shape changes**, not just numbers.

*Write your answer here (double-click to edit this cell)*

When math box $A = \begin{bmatrix} 2 & 0 \\ 0 & 3 \end{bmatrix}$ times a 2D list, it makes it bigger (**scaling**).

- **`A * v1 = [2, 0]`**: List `v1` gets 2 times longer on x-line. Y stays same.
- **`A * v2 = [0, 3]`**: List `v2` gets 3 times longer on y-line. X stays same.
- **`A * v3 = [2, 3]`**: List `v3` gets 2 times longer on x and 3 times longer on y. Point `(1,1)` goes to `(2,3)`. It turns and gets longer.
- **`A * v4 = [-4, 12]`**: Like `v3`, this list gets 2 times longer on x (from -2 to -4) and 3 times longer on y (from 4 to 12). Same way, just bigger.

### Q2 Part (iii) — NumPy Tool

Make a NumPy tool that:
- Takes **any 2×2 box** and **any 2D list**.
- Gives **new list**.
- Writes **one short line** about x and y change.

Try on box `A` and all four lists.

````python
# Input — do not modify
A = np.array([[2, 0], [0, 3]])
v1 = np.array([1, 0])
v2 = np.array([0, 1])
v3 = np.array([1, 1])
v4 = np.array([-2, 4])
````

````python
def transform_vector(matrix, vector):
    transformed_vector = matrix @ vector

    x_factor = transformed_vector[0] / vector[0] if vector[0] != 0 else ('unchanged' if transformed_vector[0] == 0 else 'infinite scaling')
    y_factor = transformed_vector[1] / vector[1] if vector[1] != 0 else ('unchanged' if transformed_vector[1] == 0 else 'infinite scaling')

    x_desc = f"x-component scaled by {x_factor:.2f}" if isinstance(x_factor, float) else f"x-component {x_factor}"
    y_desc = f"y-component scaled by {y_factor:.2f}" if isinstance(y_factor, float) else f"y-component {y_factor}"

    print(f"Vector transformed: {vector} -> {transformed_vector}. {x_desc}, {y_desc}.")
    return transformed_vector

print("--- Testing with v1 ---")
transform_vector(A, v1)
print("\n--- Testing with v2 ---")
transform_vector(A, v2)
print("\n--- Testing with v3 ---")
transform_vector(A, v3)
print("\n--- Testing with v4 ---")
transform_vector(A, v4)
````

### Q3 Part (i) — How Similar Matrix

Make the **similar matrix** with **one NumPy step** (no loops).

What math idea makes this easy?

````python
# Input — do not modify
users = np.array([
    [0.9, 0.1, 0.8, 0.2],  # User A
    [0.1, 0.9, 0.2, 0.8],  # User B
    [0.8, 0.2, 0.9, 0.1],  # User C
    [0.5, 0.5, 0.5, 0.5],  # User D
])
````

````python
similarity_matrix = users @ users.T
display(similarity_matrix)
````

**What math idea lets you do this in one line?**

The math idea is **dot product**, or **math box times** (`@` in NumPy). When you times a box `U` by its opposite `U.T`, the number at `(i, j)` is the dot product of row `i` and row `j`. This shows how similar every user is to every other user.

### Q3 Part (iii) — Bad Simple Dot Products

Senior worker says:
> *"Simple dot products are bad for similar — you must make normal."*

1. What is **bad** about simple dot products?
2. Make **NumPy example** to show problem.
3. Fix it.

> **Help:** Use user `[1.8, 0.2, 1.6, 0.4]`

*Write your answer here (double-click to edit this cell)*

1. **Problem:** Simple dot products care too much about how **big** the numbers are. If a user has big numbers, their similar score is always big, even if they like different things.

2. **Example:** User A is `[0.9, 0.1, 0.8, 0.2]`. `Big_User` is `[1.8, 0.2, 1.6, 0.4]`. `Big_User` has 2 times bigger numbers but same taste. Simple dot product says they are different or too big.

3. **Fix:** We **normalize** lists first. This makes list length 1. Then dot product shows the **angle** (cosine similar). 1 means same taste.

````python
import numpy as np

users = np.array([
    [0.9, 0.1, 0.8, 0.2],
    [0.1, 0.9, 0.2, 0.8],
    [0.8, 0.2, 0.9, 0.1],
    [0.5, 0.5, 0.5, 0.5],
])

inflated_user = np.array([1.8, 0.2, 1.6, 0.4])

raw_sim_A_C = np.dot(users[0], users[2])
raw_sim_A_inflated = np.dot(users[0], inflated_user)

print(f"Raw A, C: {raw_sim_A_C:.4f}")
print(f"Raw A, Big: {raw_sim_A_inflated:.4f}")

def normalize_vector(vector):
    norm = np.linalg.norm(vector)
    if norm == 0:
        return vector
    return vector / norm

users_normalized = np.array([normalize_vector(user) for user in users])
inflated_user_normalized = normalize_vector(inflated_user)

cosine_sim_A_C = np.dot(users_normalized[0], users_normalized[2])
cosine_sim_A_inflated = np.dot(users_normalized[0], inflated_user_normalized)

print(f"Good A, C: {cosine_sim_A_C:.4f}")
print(f"Good A, Big: {cosine_sim_A_inflated:.4f}")
````
